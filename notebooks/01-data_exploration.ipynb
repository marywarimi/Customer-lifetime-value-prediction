## Exploration Summary

**Dataset scope**
- Total transactions (rows): 541,909
- Unique customers: 4,372
- Unique invoices: 25,900
- Countries: 38
- Date range: '1/10/2011 10:04' to '9/9/2011 9:52' (373 days)

**Issues found**
1. Missing `CustomerID` in ~135,080 rows — cannot attribute these to a customer
2. Missing `Description` in 1,454 rows
3. `InvoiceDate` stored as object, not datetime
4. Negative `UnitPrice`: 2 rows, "Adjust bad debt" (StockCode B) — accounting entries, not sales
5. Negative `Quantity`: 10,624 rows — customer returns, with valid CustomerIDs
6. 33 non-numeric StockCodes, including fees (POST, DOT, BANK CHARGES, AMAZONFEE), gift cards, and admin codes
7. Duplicates: none

**Planned cleaning actions (notebook 02)**
- Drop rows with missing CustomerID (CLV requires customer attribution)
- Drop bad debt and non-product StockCodes
- Keep returns (negative quantity) so TotalAmount nets out correctly
- Convert InvoiceDate to datetime
- Create TotalAmount = Quantity × UnitPrice
