```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/Online Retail.csv")
```

```python
df.shape
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["InvoiceDate"].dtype
```

```python
df["InvoiceDate"].head()
```

```python
df["CustomerID"].isna().sum()
```

```python
df = df.dropna(subset=["CustomerID"])
```

```python
df.shape
```

```python
(df["UnitPrice"] <= 0).sum()
```

```python
df = df[df["UnitPrice"] > 0]
```

```python
df.shape
```

```python
df[~df['StockCode'].str[0].str.isdigit()]['StockCode'].unique()
```

```python
df[df['StockCode'].str.startswith('DCGS')]['Description'].unique()
```

```python
non_product_codes = ['POST', 'D', 'C2', 'M', 'BANK CHARGES', 'PADS', 'DOT', 'CRUK']

df = df[~df['StockCode'].isin(non_product_codes)]
print(df.shape)
```

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
```

```python
df.to_csv('../data/processed/cleaned_transactions.csv', index=False)
```
